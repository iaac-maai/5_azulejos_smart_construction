# Azulejos: Connecting Artisans with Digital Tools to Conserve Spanish Tile Art

Proposal draft for Bo De Roeck, Sanjana Kiran, Sneha Sachin Shenoy · Operable Futures · Smart Construction Studio, IAAC 26/27 · MIDTERMS

# Summary
•	The claim that the art of ceramic tiles is dying is partly supported by evidence. The data points to a succession and knowledge-transfer crisis in Spanish crafts. I found no study showing that a lack of technology integration is the cause, so the pitch should be framed around transmission of embodied knowledge and the labour of scaling designs. Two numbers on the current pitch deck (slide 03) also need correcting, as explained below.

# Problem
•	Spain’s tile-making knowledge sits in people’s hands. UNESCO’s 2019 Talavera inscription notes that this knowledge is mostly held by master artisans and passed on orally in workshops and families. Little of it is documented, measured or scalable. When a master retires, the gesture, the stroke logic and the way a motif is stretched across a façade go with them. 
•	A second problem is planning a large tiled façade means physically laying out full-scale paper or tiles and drawing every grid by hand. That is slow and tiring, and it puts off new entrants.

# Urgency and relevance
•	Spain had 208,600 artisans in 2022, a 2.2% fall since 2019 (about 4,700 jobs), attributed mainly to missing generational renewal and small firm size.
•	Closing workshops take with them techniques that took decades or centuries to perfect.
•	Potters in a 2026 report describe a popularity revival that contrasts with uncertainty over succession, and most are of advanced age. 
•	UNESCO’s own decision says the inscription should raise young people’s interest, and it flags the risk of cultural appropriation by outside actors. That is a direct argument for an artisan-owned library. 
•	Counter-evidence you should acknowledge: a 2024 CSIC-published paper questions whether the UNESCO safeguarding measures actually save Talavera ceramics. Also, Talavera reportedly pairs training schools with workshops to secure succession. The picture is uneven, not uniform decline. 

# The GAP
•	All sources document or diagnose existing tiles. We did not find work that turns an artisan’s gesture into scalable, façade-aware pattern geometry, and an equivalent open pattern catalogue for Spain.

# Solution/Proposal
Connecting artisans with digital tools to optimize large tile pattern scaling and conserve ceramic tile art through Human-Computer Interaction technology.

Two linked interfaces:
•	Artisan Studio: draw one cell on one sheet and see it placed live in the façade grid, with its 8 neighbours for context.
•	Azulejo Library: the authenticated 2D and 3D Rhino asset library for architects, conservators, historians and artists, built from what artisans create and contribute.

# Objectives
1.	Document artisans’ gestures, not just their outputs.
2.	Remove full-scale physical layout from façade pattern design.
3.	Make symmetry, mirroring and rotation effortless and artisan-controlled.
4.	Build an artisan-authored, rights-respecting library of 2D drawings and 3D models.
5.	Validate with real artisans in Spain.

# Key results
•	Baseline measured: time for an artisan to lay out and draw a 3×3 grid conventionally vs. with the tool
•	≥50% reduction in layout time, with at least 3 artisans testing
•	Stroke-to-scan deviation ≤1 mm at 150 mm cell size
•	Edge-continuity error between adjacent cells ≤0.5 mm
•	50+ library entries with full metadata, with artisan consent recorded for each
•	Artisan usability score (e.g. SUS) ≥70, plus qualitative interviews

# Methodology
Research-through-design with artisans as co-designers: field observation, then a baseline task, then iterative prototype cycles, then a comparative evaluation (conventional vs. tool) measuring time, accuracy and perceived effort. Interview artisans, conservators and architects separately, since their needs differ.

# Pipeline
1.	Define the façade. Width, height, module (e.g. 150 mm), joint width, shrinkage → Rhino grid.
2.	Pick a cell. The tool highlights it and shows the 8 surrounding cells as ghost context.
3.	Artisan draws on one A4 sheet. Printed ArUco/AprilTag markers at the corners let the camera map the paper to the cell by homography.
4.	Capture. Overhead camera, with an optional tablet mode.
5.	Extract strokes. Track the pen tip, record (x, y, t), smooth the curves, and fit splines.
6.	Snap into the cell. An edge-handshake check makes strokes that touch a boundary continue into the neighbouring cell.
7.	Apply operators. Mirror and rotate (the symmetry group of a square), with artisan approval on every result.
8.	View the whole. Façade-level preview in the Artisan Studio.
9.	Export. .3dm, .dxf, .svg for stencils, laser cutting or CAD, plus 3D relief models.
10.	Publish to the library with metadata and consent.

# AI integration
•	MediaPipe for hands, as you already use it
•	Stroke fitting: classical filtering and spline fitting, which is cheaper and more predictable than “AI” for this step
•	Vision model: detects symmetry and repeat structure, and asks the artisan before completing anything
•	Pattern retrieval: embedding-based matching against the library, similar in spirit to the Az Infinitum search
•	Voice: speech-to-text → command schema → Rhino via MCP. Use push-to-talk (workshops are noisy) and test Spanish commands like “espejo” and “gira 90.”
•	Guardrail: AI proposes, the artisan decides. This is the same “human approval” you already put on your Zoom Out slide.
